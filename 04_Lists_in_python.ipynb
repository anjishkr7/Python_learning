```python
#creating a list

#Empty list
print([])
#1D list and homogeneous list
print([1,2,3,4,5])
#2D
print([1,2,3,[4,5]])#->its a heterogenous list
#3D
print([[[1,2],[3,4]],[[5,6],[7,8]]])#-> its a homogeneous list
#Hetrogeneous
print([1,True,5.6,5+6j,'Hello'])
#using Type Conversion
print(list('hello'))

```

```python
#Accessing items from a list
#Indexing
l=[1,2,3,[4,5]]
#+ve
print(l[3])
#-ve
print(l[-1][0])
#2D
l1=[[[1,2],['a','b']],[3,4]]
print(l1[0][1][0])

#Slicing
print(l[0:3])
print(l[-3:])
print(l[::-1])
print(l[::2])
```

```python
#Adding items to list
#append
l=[1,2,3,4,5]
l.append(True)
print(l)
```

```python
#extend
l=[1,2,3,4,5]
l.extend([6,7,8])
l.extend('delhi')
print(l)
```

```python
#insert
l=[1,2,3,4,5]
l.insert(1,100)
l.insert(1,200)
print(l)
```

```python
#Editing items in a list
l=[1,2,3,4,5]
l[-1]=500
l[1:4]=[200,300,400]
print(l)
```

```python
#Deleting items from a list
#del
l=[1,2,3,4,5]
print(l)
del l
print(l)
```

```python
#indexing
l=[1,2,3,4,5]
del l[-1]
print(l)
```

```python
#slicing
l=[1,2,3,4,5]
del l[1:3]
print(l)
```

```python
#remove
l=[1,2,3,4,5]
l.remove(5)
print(l)
```

```python
#pop
l=[1,2,3,4,5]
l.pop(0)
print(l)
l.pop()
print(l)
```

```python
#clear
l=[1,2,3,4,5]
l.clear()
print(l)
```

```python
#Operation on list
#Aritmetic(+ and * only)
l1=[1,2,3,4,5]
l2=[6,7,8]
print(l1+l2)
print(l1*3)
```

```python
#membership
l1=[1,2,3,4,5]
l2=[1,2,3,4,[5,6]]
print(5 in l1)
print(5 not in l1)
print(5 in l2)
print(5 in l2[4])
print([5,6] in l2)
```

```python
#loops
l=[1,2,3,4,5]
for i in l:
  print(i)
```

```python
#List functions
#len/min/max/sorted
l=[1,2,3,4,5]
print(len(l))
print(min(l))
print(max(l))
print(sorted(l))
print(sorted(l,reverse=True))
```

```python
#count
l=[1,2,1,3,4,5]
l.count(1)
```

```python
#index
l=[1,2,3,4,5]
l.index(3)
```

```python
#reverse
l=[1,2,3,4,5]
l.reverse()#it reverses original list
print(l)
```

```python
#sort (vs sorted)
l=[9,8,7,6]
print(l)
print(sorted(l))#sorted doesnt sort the original list but sort does
print(l)
l.sort()
print(l)
```

```python
#copy-> shallow copy
l=[1,2,3,4]
print(l)
print(id(l))
l1=l.copy()
print(l1)
print("Address: ",id(l1))
```

```python
#list comprehension
#Add 1 to 10 numbers to a list
l=[]
for i in range(1,11):
  l.append(i)
print(l)
#by list comprehension
l1=[i for i in range(1,11)]
print(l1)
```

```python
#scalar multi on a vector
v=[2,3,4]
s=-3
x=[i*s for i in v]
print(x)
```

```python
#add squares
l=[1,2,3,4,5]
x=[i**2 for i in l]
print(x)
```

```python
#print all numbers divisible by 5 in the range of 1 to 50
x=[i for i in range(1,51) if i%5==0]
print(x)
```

```python
#find languages which starts with letter p
languages=['java','python','php','c','javascript']
x=[language for language in languages if language.startswith('p')]
print(x)
```

```python
#nested if with list comprehension
basket=['apple','guava','cherry','banana']
my_fruits=['apple','kiwi','grapes','banana']
#add new list from my_fruits and items if the fruits exists in basket and also stats with 'a'
x=[fruit for fruit in my_fruits if fruit in basket if fruit.startswith('a')]
print(x)
```

```python
##3 x 3 matrix
x=[[i*j for i in range(1,4)] for j in range(1,4)]
print(x)
```

```python
#cartisan prod
l1=[1,2,3,4]
l2=[5,6,7,8]
[i*j for i in l1 for j in l2]
```

```python
#2 ways to traverse a list
#itemwise
l=[1,2,3,4]
for i in l:
  print(i)
```

```python
#indexwise
l=[1,2,3,4]
for i in range(0,len(l)):
  print(l[i])
```

```python
# zip function
l1=[1,2,3,4]
l2=[1,-2,-3,-4]
# list(zip(l1,l2))
[i+j for i,j in zip(l1,l2)]
```
