#LOOPS IN PYTHON

```python
#while loop
number=int(input("Enter a no: "))
i=1
while i<11:
  print(number*i)
  i+=1
```

```python
# while loop with else
x=1
while x<3:
  print(x)
  x+=1
else:
  print('limit crossed')
```

```python
# Guessing game
import random
randomNum=random.randint(1,100)
guessedNum=int(input("Guess the number: "))
count=1
while guessedNum!=randomNum:
  if guessedNum<randomNum:
    print('wrong! Guess higher')
  else:
    print('wrong! guess lower')
  guessedNum=int(input("Guess the number: "))
  count+=1
else:
  print('yeah, You guessed it write and the number is',randomNum,',\nand u have taken',count,'attempts')
```

```python
n=int(input("Enter n "))
fact=1
result=0
for i in range(1,n+1):
  fact=fact*i
  result+=i/fact
print(result)
```

#NESTED LOOPS


```python
#unique Pairs
for i in range(1,5):
  for j in range(1,5):
    print(i,j)
```

```python
#pattern 1
n=int(input("Enter the number of rows: "))
for i in range(1,n+1):
  for j in range(i):
    print('*',end='')
  print()
```

```python
#pattern 2
n=int(input("Enter the number of rows: "))
for i in range(1,n+1):
  for j in range(1,i+1):
    print(j,end="")
  for k in range(i-1,0,-1):
    print(k,end="")
  print()
```

#Loop control statments

```python
#break
for i in range(1,10):
  if i==5:
    break
  print(i)
```

```python
lower_limit=int(input("Enter lower limit: "))
upper_limit=int(input("Enter upper limit: "))
for i in range(lower_limit,upper_limit):
  for j in range(2,i):
    if i%j==0:
      break
  else:
    print(i)
```

```python
#continue
for i in range(10):
  if i==5:
    continue
  print(i)
```
